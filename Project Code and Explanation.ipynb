Lets compare the distribution of the sample and traing data sets.  

We will first see if the returns are normal and if there are any outliters by using the historgram, box plot, and the Q-Q plot.  
The **Q-Q plot** can be used to check if the data set has a normal distribution.  
The points on the Q-Q plot are created by pairing the quantiles from both distributions.  
If the data points are close to the linear line, it means the data set is close to a normal distribution.  

We will also use the **Jarque-Bera Test** from the stats library.  
The Bera test basically tells us if the data is normaly distributed or not by computing the p-value, which gives a numerical metric to determine normality.   
If p-value is greater than 0.05 it means it is a normal distribution.

```python
from scipy import stats

# Prepare data
AAPL_Train = helper.getTrainData('AAPL')
AAPL_Train.columns = [col.replace('AAPL_', '') for col in AAPL_Train.columns]
AAPL_Train_Adj = helper.adjust_all_prices(AAPL_Train)
AAPL_Train_Adj['Return'] = AAPL_Train_Adj['Adj_Close'].pct_change()

AAPL_Sample = helper.getSampleData('AAPL')
AAPL_Sample.columns = [col.replace('AAPL_', '') for col in AAPL_Sample.columns]
AAPL_Sample_Adj = helper.adjust_all_prices(AAPL_Sample)
AAPL_Sample_Adj['Return'] = AAPL_Sample_Adj['Adj_Close'].pct_change()

# Create comparison plots
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Plot 1: Histogram Comparison
returns_train_pct = AAPL_Train_Adj['Return'] * 100
returns_sample_pct = AAPL_Sample_Adj['Return'] * 100

_=axes[0, 0].hist(returns_train_pct, bins=100, edgecolor='black', alpha=0.5, 
                density=True, label='Train (2000-2016)', color='blue')
_=axes[0, 0].hist(returns_sample_pct, bins=30, edgecolor='black', alpha=0.5, 
                density=True, label='Sample (2017)', color='orange')

# Normal distribution for train
mu_train, sigma_train = returns_train_pct.mean(), returns_train_pct.std()
x_train = np.linspace(returns_train_pct.min(), returns_train_pct.max(), 100)
_=axes[0, 0].plot(x_train, stats.norm.pdf(x_train, mu_train, sigma_train), 
                'b--', linewidth=2, label='Normal (Train)', alpha=0.7)

# Normal distribution for sample
mu_sample, sigma_sample = returns_sample_pct.mean(), returns_sample_pct.std()
x_sample = np.linspace(returns_sample_pct.min(), returns_sample_pct.max(), 100)
_=axes[0, 0].plot(x_sample, stats.norm.pdf(x_sample, mu_sample, sigma_sample), 
                'r--', linewidth=2, label='Normal (Sample)', alpha=0.7)

_=axes[0, 0].set_title('Distribution of Daily Returns: Train vs Sample', fontsize=12, fontweight='bold')
_=axes[0, 0].set_xlabel('Return (%)')
_=axes[0, 0].legend()
_=axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Q-Q Plot Comparison (side by side in same plot)
from scipy.stats import probplot

# Train Q-Q
train_qq = probplot(AAPL_Train_Adj['Return'].dropna(), dist="norm")
_=axes[0, 1].scatter(train_qq[0][0], train_qq[0][1], alpha=0.5, s=10, label='Train', color='blue')
_=axes[0, 1].plot(train_qq[0][0], train_qq[1][1] + train_qq[1][0]*train_qq[0][0], 
                'b--', linewidth=2, alpha=0.7)

# Sample Q-Q
sample_qq = probplot(AAPL_Sample_Adj['Return'].dropna(), dist="norm")
_=axes[0, 1].scatter(sample_qq[0][0], sample_qq[0][1], alpha=0.5, s=10, label='Sample', color='orange')
_=axes[0, 1].plot(sample_qq[0][0], sample_qq[1][1] + sample_qq[1][0]*sample_qq[0][0], 
                'r--', linewidth=2, alpha=0.7)

_=axes[0, 1].set_title('Q-Q Plot: Train vs Sample', fontsize=12, fontweight='bold')
_=axes[0, 1].set_xlabel('Theoretical Quantiles')
_=axes[0, 1].set_ylabel('Sample Quantiles')
_=axes[0, 1].legend()
_=axes[0, 1].grid(True, alpha=0.3)

# Plot 3: Returns Over Time (both datasets)
_=axes[1, 0].plot(AAPL_Train_Adj.index, returns_train_pct, linewidth=0.5, alpha=0.8, 
                label='Train (2000-2016)', color='blue')
_=axes[1, 0].plot(AAPL_Sample_Adj.index, returns_sample_pct, linewidth=1.5, alpha=0.8, 
                label='Sample (2017)', color='orange')
_=axes[1, 0].axhline(y=0, color='red', linestyle='--', alpha=0.5)
_=axes[1, 0].set_title('Daily Returns Over Time: Train vs Sample', fontsize=12, fontweight='bold')
_=axes[1, 0].set_ylabel('Return (%)')
_=axes[1, 0].legend()
_=axes[1, 0].grid(True, alpha=0.3)

# Plot 4: Box Plot Comparison
box_data = [returns_train_pct.dropna(), returns_sample_pct.dropna()]
bp = axes[1, 1].boxplot(box_data, labels=['Train\n(2000-2016)', 'Sample\n(2017)'], 
                         patch_artist=True)
_=bp['boxes'][0].set_facecolor('blue')
_=bp['boxes'][0].set_alpha(0.5)
_=bp['boxes'][1].set_facecolor('orange')
_=bp['boxes'][1].set_alpha(0.5)
_=axes[1, 1].set_title('Box Plot: Train vs Sample', fontsize=12, fontweight='bold')
_=axes[1, 1].set_ylabel('Return (%)')
_=axes[1, 1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# Statistical Comparison
print("\nTRAIN DATA (2000-2016):")
print(f"  Observations: {len(AAPL_Train_Adj['Return'].dropna())}")
print(f"  Mean:   {AAPL_Train_Adj['Return'].mean()*100:.4f}%")
print(f"  Std:    {AAPL_Train_Adj['Return'].std()*100:.4f}%")
print(f"  Min:    {AAPL_Train_Adj['Return'].min()*100:.4f}%")
print(f"  Max:    {AAPL_Train_Adj['Return'].max()*100:.4f}%")
print(f"  Skew:   {AAPL_Train_Adj['Return'].skew():.4f}")
print(f"  Kurt:   {AAPL_Train_Adj['Return'].kurtosis():.4f}")

print("\nSAMPLE DATA (2017):")
print(f"  Observations: {len(AAPL_Sample_Adj['Return'].dropna())}")
print(f"  Mean:   {AAPL_Sample_Adj['Return'].mean()*100:.4f}%")
print(f"  Std:    {AAPL_Sample_Adj['Return'].std()*100:.4f}%")
print(f"  Min:    {AAPL_Sample_Adj['Return'].min()*100:.4f}%")
print(f"  Max:    {AAPL_Sample_Adj['Return'].max()*100:.4f}%")
print(f"  Skew:   {AAPL_Sample_Adj['Return'].skew():.4f}")
print(f"  Kurt:   {AAPL_Sample_Adj['Return'].kurtosis():.4f}")

# Normality tests
jb_train_stat, jb_train_p = stats.jarque_bera(AAPL_Train_Adj['Return'].dropna())
print(f"\nTrain - Jarque-Bera Test:")
print(f"  Statistic: {jb_train_stat:.4f}")
print(f"  P-value:   {jb_train_p:.6f}")
print(f"  Result:    {'NOT normal' if jb_train_p < 0.05 else 'Normal'} (reject H0 if p < 0.05)")

jb_sample_stat, jb_sample_p = stats.jarque_bera(AAPL_Sample_Adj['Return'].dropna())
print(f"\nSample - Jarque-Bera Test:")
print(f"  Statistic: {jb_sample_stat:.4f}")
print(f"  P-value:   {jb_sample_p:.6f}")
print(f"  Result:    {'NOT normal' if jb_sample_p < 0.05 else 'Normal'} (reject H0 if p < 0.05)")
```

Moreover we will introduce some and features, metrics, and calculations like lagged features, Autocorrelation Function (ACF), and ARCH effects.  

**Lagged features** allow us to use past values of a variable as predictors for current/future values.  
This ensures that we do not accidentally peak into the future and cheat.  

The **ACF** measures how correlated a time series is with lagged versions of itself. The calculation step is the following:  
1. Calculate the sample mean of the time series.   
2. Calculate the sample variance , which is the sum of the squared deviations from the mean, divided by the number of observations.   
3. Calculate the sample autocovariance for each lag.  For a specific lag, multiply the difference of each point and the mean by the difference of the point periods prior and the mean.  
4. Sum these products and divide by the number of observations.  
5. Calculate the sample ACF for each lag by dividing the sample autocovariance at that lag by the sample variance.   

A high ACF essentially means the time series data has high memory and thus may be better for prediction.  

The **ARCH (Autoregressive Conditional Heteroskedasticity) effect** describes volatility clustering.  
If ARCH effect is present, then volatility is not constant.  
The p-value of the ARCH indicates the presence of ARCH effects.  
It seems that typically p-value less than 0.05 is reasonable to conclude that there is evidence of ARCH effects and a p-value of less than 0.001 indicates a strong ARCH effect.

```python

from keras import Sequential
from keras.layers import Input
from keras.layers import Dense, Dropout
from keras.models import Sequential
from keras.layers import Reshape, Conv1D, Dropout, GlobalMaxPooling1D, Dense
set_random_seed(42)
np.random.seed(42)
os.environ['PYTHONHASHSEED'] = str(42)

# 1. Baseline FCNN
baseline = Sequential([
    Input(shape=(17,)),  
    Dense(32, activation='relu'),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dropout(0.2),
    Dense(1)
], name='baseline_fcnn')
baseline.compile(optimizer='adam', loss='mse', metrics=['mae'])
# Parameters: ~1,100

# 2. Deep FCNN
deep = Sequential([
    Input(shape=(17,)),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dropout(0.2),
    Dense(1)
], name='deep_fcnn')
deep.compile(optimizer='adam', loss='mse', metrics=['mae'])
# Parameters: ~3,500

# 3. Wide FCNN
wide = Sequential([
    Input(shape=(17,)),
    Dense(128, activation='relu'),
    Dropout(0.2),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(1)
], name='wide_fcnn')
wide.compile(optimizer='adam', loss='mse', metrics=['mae'])
# Parameters: ~10,000

# 4. Regularized FCNN (higher dropout + L2)
from keras.regularizers import l2
regularized = Sequential([
    Input(shape=(17,)),
    Dense(32, activation='relu', kernel_regularizer=l2(0.01)),
    Dropout(0.3),
    Dense(16, activation='relu', kernel_regularizer=l2(0.01)),
    Dropout(0.3),
    Dense(1)
], name='regularized_fcnn')
regularized.compile(optimizer='adam', loss='mse', metrics=['mae'])
# Parameters: ~1,100 (same as baseline, but more regularization)

# 5. 1D CNN (treat features as sequence)
# Reshape input: (batch, 17, 1) - treat 17 features as sequence
cnn = Sequential([
    Input(shape=(17,)),
    Reshape((17, 1)),
    Conv1D(32, kernel_size=3, activation='relu', padding='same'),
    Dropout(0.2),
    Conv1D(16, kernel_size=3, activation='relu', padding='same'),
    GlobalMaxPooling1D(),
    Dense(8, activation='relu'),
    Dropout(0.2),
    Dense(1)
], name='cnn_1d')
cnn.compile(optimizer='adam', loss='mse', metrics=['mae'])
# Parameters: ~1,500

non_sequential_models = {
    'baseline_fcnn_32_16': baseline,
    'deep_fcnn_64_32_16': deep,
    'wide_fcnn_128_64': wide,
    'regularized_fcnn': regularized,
    'cnn_1d': cnn
}

results = helper.compare_models(non_sequential_models, AAPL_data, plot=True)
```

GRU (Gated Recurrent Unit) is a type of recurrent neural network (RNN) that excels at handling sequential data like text or time series by using "gates" to control information flow.  
It is basically a more streamlined version of LSTM and uses two gates, the update gate and the reset gate, to capture both long-term and short-term dependencies more efficiently.

TCN (Temporal Convolutional Network) is a deep learning architecture for sequence and time-series data that uses causal convolutions to capture temporal dependencies.   
Unlike traditional recurrent networks like LSTMs, TCNs can process sequences in parallel, handle long-term dependencies efficiently, and avoid vanishing/exploding gradient problems.  

The sequence models will use the same loss and optimizer and parameters as the non sequence models we defined previously for a fiar comparison.  
Moreover, we will use a modest dropout of 0.2 for all sequence models.  
For TCN, we will have an increasing dilitation layer.  
Dilitation factor controls the spacing between the input samples used by the convolutional filter.  
The main goal of a TCN is to build up a very large receptive field—the length of the input history that one output prediction can "see" with a relatively shallow network.  
By doubling the dilation rate in each layer, the receptive field grows exponentially with the depth of the network.   
This means a shallow TCN with just a few layers can capture very long-range dependencies, long-term memory, which would give an advantage compared to Recurrent Neural Networks like LSTMs and GRUs.  
We have acual padding for TCN to prevent future information leakage.  

The data for the sequence models will have the adjusted close, open, high, low and volume since sequence models can take adjusted raw features unlike non sequence models.  
The lookback will be 60 days since one of the features is 60 day volatility (vol_60), which has the longest time period of all the features.  

```python
AAPL_train_seq = helper.prepare_features('AAPL', use_train=True, is_sequence=True)
AAPL_sample_seq = helper.prepare_features('AAPL', use_train=False, is_sequence=True)

print(f"  Columns: {list(AAPL_train_seq.columns)}")
print(f"  Train shape: {AAPL_train_seq.shape}")
print(f"  Sample shape: {AAPL_sample_seq.shape}")

def create_sequences(df, lookback=50):
    """
    Create sequences from features DataFrame
    Returns:
    X : array (n_samples, lookback, n_features)
    y : array (n_samples,)
    """
    import numpy as np
    
    # Separate features and target
    # Return is both a feature (in sequences) and target (next day)
    feature_cols = [col for col in df.columns if col != 'Return']
    
    X_seq = []
    y_targets = []
    
    for i in range(lookback, len(df)):
        # Sequence: past lookback days of features
        seq = df[feature_cols].iloc[i-lookback:i].values
        # Target: current day's return
        target = df['Return'].iloc[i]
        
        X_seq.append(seq)
        y_targets.append(target)
    
    return np.array(X_seq), np.array(y_targets)

lookback = 60

X_train_seq, y_train_seq = create_sequences(AAPL_train_seq, lookback)
X_sample_seq, y_sample_seq = create_sequences(AAPL_sample_seq, lookback)

print(f"  Train: {X_train_seq.shape} (samples, timesteps, features)")
print(f"  Sample: {X_sample_seq.shape}")

AAPL_seq_data = helper.prepare_sequences_for_training(
    X_train_seq, y_train_seq,
    X_sample_seq, y_sample_seq,
    val_ratio=0.15
)

n_features = X_train_seq.shape[2]
print(f"\n Number of features: {n_features}")
```

```python
from keras.layers import LSTM, GRU, GlobalAveragePooling1D, MultiHeadAttention, LayerNormalization
set_random_seed(42)
np.random.seed(42)
os.environ['PYTHONHASHSEED'] = str(42)

# 1. LSTM
lstm = Sequential([
    Input(shape=(lookback, n_features)),
    LSTM(50, return_sequences=True),
    Dropout(0.2),
    LSTM(25),
    Dropout(0.2),
    Dense(1)
], name='lstm')
lstm.compile(optimizer='adam', loss='mse', metrics=['mae'])
print(f"LSTM: {lstm.count_params():,} parameters")

# 2. GRU
gru = Sequential([
    Input(shape=(lookback, n_features)),
    GRU(50, return_sequences=True),
    Dropout(0.2),
    GRU(25),
    Dropout(0.2),
    Dense(1)
], name='gru')
gru.compile(optimizer='adam', loss='mse', metrics=['mae'])
print(f"GRU: {gru.count_params():,} parameters")

# 3. TCN (Temporal Convolutional Network)
tcn = Sequential([
    Input(shape=(lookback, n_features)),
    # Convolutions with increasing dilation rates
    Conv1D(32, kernel_size=3, dilation_rate=1, padding='causal', activation='relu'),
    Dropout(0.1),
    Conv1D(32, kernel_size=3, dilation_rate=2, padding='causal', activation='relu'),
    Conv1D(32, kernel_size=3, dilation_rate=4, padding='causal', activation='relu'),
    Dropout(0.1),
    Conv1D(32, kernel_size=3, dilation_rate=8, padding='causal', activation='relu'),
    GlobalAveragePooling1D(),
    Dense(16, activation='relu'),
    Dense(1)
], name='tcn')
tcn.compile(optimizer='adam', loss='mse', metrics=['mae'])
print(f"TCN: {tcn.count_params():,} parameters")

# 4. Transformer
def create_transformer(lookback, n_features, num_heads=4, ff_dim=32):
    """
    Simple Transformer for time series regression
    Uses multi-head self-attention to learn temporal patterns
    """
    inputs = Input(shape=(lookback, n_features))
    
    # Multi-head self-attention
    attn_output = MultiHeadAttention(
        num_heads=num_heads, 
        key_dim=n_features,
        dropout=0.1
    )(inputs, inputs)
    attn_output = Dropout(0.2)(attn_output)
    
    # Add & Norm (residual connection)
    x = LayerNormalization(epsilon=1e-6)(inputs + attn_output)
    
    # Feed-forward network
    ff_output = Dense(ff_dim, activation='relu')(x)
    ff_output = Dropout(0.2)(ff_output)
    ff_output = Dense(n_features)(ff_output)
    
    # Add & Norm (residual connection)
    x = LayerNormalization(epsilon=1e-6)(x + ff_output)
    
    # Global pooling and output
    x = GlobalAveragePooling1D()(x)
    x = Dense(16, activation='relu')(x)
    x = Dropout(0.2)(x)
    outputs = Dense(1)(x)
    
    model = keras.Model(inputs=inputs, outputs=outputs, name='transformer')
    return model

transformer = create_transformer(lookback, n_features)
transformer.compile(optimizer='adam', loss='mse', metrics=['mae'])

print(f"Transformer: {transformer.count_params():,} parameters")

sequence_models = {
    'lstm': lstm,
    'gru': gru,
    'tcn': tcn,
    'transformer': transformer
}

seq_results = helper.compare_sequence_models(
    sequence_models, 
    AAPL_seq_data, 
    plot=True
)
```

Based on the results, it seems that transformer, TCN, and gru have equally high directional accuracies, which are higher than the directional accuracies acheives by FCCN or CNN models.

## Error Analysis

We will first do error analysis of the TCN model we defined above on the validation and test data.  
We will plot the histogram of the residual distribution, predictions vs actual, errors over time, Q-Q plot, error return by magnitude, and residual autocorrelation.  

The **histogram of the residual distribution** is basically the histogram of the errors.  
If the histogram is centered near zero, the model has no bias.  
If the shape of the distribution is normal, the model does not produce large errors (outliers) as much. If the distributino is skewed it indicates the errors are not symmetrical.  

The **prediction vs actuals scatter plot** has a red line which basically indicates a perfect prediction.  
A low correlation indicates that there is little predictive power of the model.  

The **errors over time plot** should show errors that are random and centered around the horizontal zero line.  
For example, the moving average error should be oscillating around zero.  
If it does not, it would indicate time dependent bias.  

The **Q-Q plot** compares quantiles of residuals with the quantiles of the theoretical normal distribution.  
If the tail of the Q-Q plot curves, it indicates heavier tails than normal and hows errors are more extreme.  

The **error by return magnitude** shows the Mean Absolute Error (MAE) based on the actual return.  
Ideally the error bars of different magnitudes should have around the same height which would mean consistent predictive accuracy accross different volatility levels.  

The **residual autocorrelation plot** is the ACF plot.  
Ideally the errors or residuals should be independent showing uncorrelation with time.  
The dash lines will represent the 95% confidence interval.  
Error bars extending outside the confidence interval line means that the model missed a linear time series pattern and that the error should be predictable based on previous errors.

Lets adjust the TCN model.  
First we will add batch normalization to each layer.  
Batch normalization standardizes the inputs to a layer for each mini-batch during training which stabilizes and accelerates neural network training.  
Since neural networks are sensitive to scale, we will add batch normalization to each layer.  

We will maintain the kernel size to 3 as it sems that is the most common kernel size that results in stability of the TCN model.  
It also seems that maintaing consistent kernel size in all layers is crucial and not doing so will hurt the optimization, introduce instability, and complicates **repective field calculation**.  
Moreover the dilitation rate should ideally be increasing exponentially in powers of 2.  
This also has to do with the receptive field calculation.  
TCN, a sequence model, can bascially create lagged features internally.  
The receptive field calculation is the folllowing: Receptive field (RF) increase = (kernel size - 1)* dilitation rate.  
For example, a TCN with 3 layers each with 1, 2, 4 dilitation rate with 3 kernel sizes will have the 1st layer covers up to 2 lag time ((3-1)*1), 2nd layer covers up to 6 lag time, 3rd layer covers up to 12 lag time.  
This ensures that all lag times up to 12 are covered throughout the model.  
Finally, we will use a dropout of 0.1 every other layer.

**I just realized I may have made a mistake for the tcn models used for the error analysis.**  
The lookback we used is 60 days, but for our tcn models (which had a max dilitation rate of 8 with kernel size 3) has a Receptive field increase of only (3-1)*8 = 16 days.

This means of the 60 day lookback, the model only really utilized 16 day lag which is an issue.

```python
def create_sequences(df, lookback=50):
    """
    Create sequences from features DataFrame
    Returns:
    X : array (n_samples, lookback, n_features)
    y : array (n_samples,)
    """
    import numpy as np
    
    # Separate features and target
    # Return is both a feature (in sequences) and target (next day)
    feature_cols = [col for col in df.columns if col != 'Return']
    
    X_seq = []
    y_targets = []
    
    for i in range(lookback, len(df)):
        # Sequence: past lookback days of features
        seq = df[feature_cols].iloc[i-lookback:i].values
        # Target: current day's return
        target = df['Return'].iloc[i]
        
        X_seq.append(seq)
        y_targets.append(target)
    
    return np.array(X_seq), np.array(y_targets)
    
lookback_30 = 30

AAPL_train_final = helper.prepare_features('AAPL', use_train=True, is_sequence=True)
SPY_train_final = helper.prepare_features('SPY', use_train=True, is_sequence=True)
AAPL_sample_final = helper.prepare_features('AAPL', use_train=False, is_sequence=True)
SPY_sample_final = helper.prepare_features('SPY', use_train=False, is_sequence=True)

# Combine
AAPL_SPY_train_final = AAPL_train_final.join(SPY_train_final, rsuffix='_SPY', how='outer')
AAPL_SPY_sample_final = AAPL_sample_final.join(SPY_sample_final, rsuffix='_SPY', how='outer')

X_train_seq_final, y_train_seq_final = create_sequences(AAPL_SPY_train_final, lookback_30)
X_sample_seq_final, y_sample_seq_final = create_sequences(AAPL_SPY_sample_final, lookback_30)

AAPL_SPY_data_final = helper.prepare_sequences_for_training(X_train_seq_final, y_train_seq_final,X_sample_seq_final, y_sample_seq_final, val_ratio=0.00)

# Fix Y shape in case not 1D
AAPL_SPY_data_final['y_train'] = np.asarray(AAPL_SPY_data_final['y_train']).flatten()
AAPL_SPY_data_final['y_test'] = np.asarray(AAPL_SPY_data_final['y_test']).flatten()

import pickle
with open('feature_scalers.pkl', 'wb') as f:
    pickle.dump(AAPL_SPY_data_final['feature_scalers'], f)

n_features_final = X_train_seq_final.shape[2]

lookback_30 = 30

# final model is copy of tcn_9
final_model = Sequential([
    Input(shape=(lookback_30, n_features_final)),
    Conv1D(32, kernel_size=3, dilation_rate=1, padding='causal', activation='relu'),
    BatchNormalization(),
    Dropout(0.1),
    Conv1D(32, kernel_size=3, dilation_rate=2, padding='causal', activation='relu'),
    BatchNormalization(),
    Conv1D(32, kernel_size=3, dilation_rate=4, padding='causal', activation='relu'),
    BatchNormalization(),
    Dropout(0.1),
    Conv1D(32, kernel_size=3, dilation_rate=8, padding='causal', activation='relu'),
    BatchNormalization(),
    GlobalAveragePooling1D(),
    Dense(16, activation='relu'),
    Dense(1)
], name='final_model')
final_model.compile(optimizer='adam', loss='mae', metrics=['mae'])

history = final_model.fit(
    AAPL_SPY_data_final['X_train'], 
    AAPL_SPY_data_final['y_train'],
    epochs=100,
    batch_size=32,
    verbose=0
)
```
